```python
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

x=df.drop('Class', axis=1)
y=df['Class']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(probability=True, random_state=42))
])

param_grid = [
    {
        'scaler': [StandardScaler(), RobustScaler()],
        'svm__kernel': ['rbf'],
        'svm__C': [1, 5, 10, 20, 50, 100],
        'svm__gamma': ['scale', 'auto', 0.01, 0.05, 0.1, 0.2]
    },
    {
        'scaler': [StandardScaler(), RobustScaler()],
        'svm__kernel': ['linear'],
        'svm__C': [0.1, 1, 10, 50]
    }
]

cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid_search = GridSearchCV(pipeline, param_grid, cv=cv, scoring='accuracy', n_jobs=-1)
grid_search.fit(x_train, y_train)

print("Best Parameters :", grid_search.best_params_)
print("Best CV Accuracy:", f"{grid_search.best_score_ * 100:.2f}%")

```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, accuracy_score

best_model = grid_search.best_estimator_
y_pred = best_model.predict(x_test)
y_proba = best_model.predict_proba(x_test)[:, 1]

print(f"Test Accuracy : {accuracy_score(y_test, y_pred) * 100:.2f}%")
print(f"ROC-AUC Score : {roc_auc_score(y_test, y_proba):.4f}\n")
print("--- Classification Report ---")
print(classification_report(y_test, y_pred, target_names=['Çerçevelik (0)', 'Ürgüp Sivrisi (1)']))


cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Çerçevelik', 'Ürgüp Sivrisi'],
            yticklabels=['Çerçevelik', 'Ürgüp Sivrisi'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()
```
